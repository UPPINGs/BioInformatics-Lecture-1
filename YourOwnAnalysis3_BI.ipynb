```python
import pandas as pd

pileup = pd.read_csv('CLIP-let7g-gene.pileup', sep='\t', names=['chrom', 'pos', '_ref', 'count', 'basereads', 'quals'])
pileup.tail()
```

```python
import re
toremove = re.compile('[<>$*#^]')
pileup['matches'] = pileup['basereads'].apply(lambda x: toremove.sub('', x))
```

```python
pileup[['chrom', 'pos', 'matches']]
```

```python
import numpy as np
from collections import Counter
import math

def calculate_entropy(sequence):
    if not sequence:
        return 0
    counts = Counter(sequence)
    total = sum(counts.values())
    probs = [count / total for count in counts.values()]
    return -sum(p * math.log2(p) for p in probs if p > 0)

# 1 & 2. 각 위치별 Shannon entropy 계산
pileup['entropy'] = pileup['matches'].apply(calculate_entropy)

# 3. bedGraph format (chrom, start, end, value) 생성
# bedGraph는 0-based start, 1-based end를 사용하므로 pos-1을 start로 잡습니다.
bedgraph = pileup[['chrom', 'pos', 'entropy']].copy()
bedgraph['start'] = bedgraph['pos'] - 1
bedgraph['end'] = bedgraph['pos']
bedgraph = bedgraph[['chrom', 'start', 'end', 'entropy']]

# 결과 저장
output_filename = '/content/drive/MyDrive/Colab_Notebooks/BI_lecture1/mission3_bedgraph/mirlet7g.bedgraph'
bedgraph.to_csv(output_filename, sep='\t', header=False, index=False)
print(f'{output_filename} has been created.')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

print("1. 위치별 Read Coverage 및 Entropy 상대 위치 분석 시작...")

try:
    df_to_plot = pileup.copy()
except NameError:
    try:
        df_to_plot = bedgraph.copy()
    except NameError:
        print("⚠️ pileup 또는 bedgraph 데이터프레임을 찾을 수 없습니다. 위쪽 셀들이 정상 실행되었는지 확인해주세요.")

if 'df_to_plot' in locals():
    # 1. 절대 좌표(pos 또는 start)를 시작점 기준 상대 좌표(0, 1, 2...)로 변환
    if 'pos' in df_to_plot.columns:
        df_to_plot['rel_pos'] = df_to_plot['pos'] - df_to_plot['pos'].min()
    elif 'start' in df_to_plot.columns:
        df_to_plot['rel_pos'] = df_to_plot['start'] - df_to_plot['start'].min()

    # count 컬럼이 없는 경우(bedgraph 등)를 대비한 예외 처리
    has_count = 'count' in df_to_plot.columns

    if has_count:
        # 리드 수(count)를 비율(%)로 정규화
        total_count = df_to_plot['count'].sum()
        df_to_plot['coverage_pct'] = (df_to_plot['count'] / total_count) * 100

    # 2. 시각화 (2단 그래프)
    fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)

    # 상단: Coverage Profile (count가 있는 경우만 그림)
    if has_count:
        axes[0].plot(df_to_plot['rel_pos'], df_to_plot['coverage_pct'], color='#3498DB', marker='o', linewidth=2)
        axes[0].set_ylabel('Normalized Coverage (%)', fontsize=11)
        axes[0].set_title('Read Coverage Profile along pre-miRNA', fontsize=12, fontweight='bold')
    else:
        axes[0].text(0.5, 0.5, 'Coverage data (count) not available in this dataframe', ha='center', va='center')
        axes[0].set_title('Read Coverage Profile', fontsize=12)

    axes[0].grid(True, linestyle=':', alpha=0.6)

    # 하단: Entropy Profile
    axes[1].plot(df_to_plot['rel_pos'], df_to_plot['entropy'], color='#E74C3C', marker='s', linewidth=1.5)
    axes[1].set_title('Shannon Entropy Profile', fontsize=12, fontweight='bold')
    axes[1].set_xlabel("Relative position from 5' end (nt)", fontsize=11)
    axes[1].set_ylabel('Shannon Entropy', fontsize=11)
    axes[1].grid(True, linestyle=':', alpha=0.6)

    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
from collections import Counter
import math

def calculate_entropy(sequence):
    if not sequence:
        return 0
    counts = Counter(sequence)
    total = sum(counts.values())
    probs = [count / total for count in counts.values()]
    return -sum(p * math.log2(p) for p in probs if p > 0)

# 1 & 2. 각 위치별 Shannon entropy 계산
pileup['entropy'] = pileup['matches'].apply(calculate_entropy)

# 3. bedGraph format (chrom, start, end, value) 생성
# bedGraph는 0-based start, 1-based end를 사용하므로 pos-1을 start로 잡습니다.
bedgraph = pileup[['chrom', 'pos', 'entropy']].copy()
bedgraph['start'] = bedgraph['pos'] - 1
bedgraph['end'] = bedgraph['pos']
bedgraph = bedgraph[['chrom', 'start', 'end', 'entropy']]

# 결과 저장
output_filename = '/content/drive/MyDrive/Colab_Notebooks/BI_lecture1/mission3_bedgraph/mirlet7d.bedgraph'
bedgraph.to_csv(output_filename, sep='\t', header=False, index=False)
print(f'{output_filename} has been created.')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 현재 분석 중인 miRNA가 negative strand인지 확인 (Mirlet7d, Mirlet7f-1 등)
# 이 예제에서는 유동적으로 처리하기 위해 좌표 계산 방식을 수정합니다.

print("1. 위치별 Read Coverage 및 Entropy 상대 위치 분석 시작 (Strand 반영)...")

try:
    df_to_plot = pileup.copy()
except NameError:
    print("⚠️ pileup 데이터프레임을 찾을 수 없습니다.")

if 'df_to_plot' in locals():
    # 1. Strand 판별 및 상대 좌표 계산
    # Mirlet7g (+) : pos - min(pos)
    # Mirlet7d/f-1 (-) : max(pos) - pos

    # 여기서는 간단히 좌표 역전 여부를 선택할 수 있게 하거나,
    # 파일명이나 데이터를 통해 자동 판별 로직을 넣을 수 있습니다.
    # 여기서는 'pos'의 최대값에서 빼는 방식을 적용하여 5'->3' 방향을 맞춥니다.

    is_negative_strand = True # Mirlet7d, Mirlet7f-1의 경우 True

    if is_negative_strand:
        df_to_plot['rel_pos'] = df_to_plot['pos'].max() - df_to_plot['pos']
    else:
        df_to_plot['rel_pos'] = df_to_plot['pos'] - df_to_plot['pos'].min()

    has_count = 'count' in df_to_plot.columns
    if has_count:
        total_count = df_to_plot['count'].sum()
        df_to_plot['coverage_pct'] = (df_to_plot['count'] / total_count) * 100

    # 2. 시각화
    fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)

    if has_count:
        # 선의 흐름을 보기 위해 rel_pos 기준으로 정렬
        df_sorted = df_to_plot.sort_values('rel_pos')
        axes[0].plot(df_sorted['rel_pos'], df_sorted['coverage_pct'], color='#3498DB', marker='o', linewidth=2)
        axes[0].set_ylabel('Normalized Coverage (%)', fontsize=11)
        axes[0].set_title('Read Coverage Profile (5\' -> 3\' direction)', fontsize=12, fontweight='bold')

    axes[0].grid(True, linestyle=':', alpha=0.6)

    # 하단: Entropy Profile
    df_sorted = df_to_plot.sort_values('rel_pos')
    axes[1].plot(df_sorted['rel_pos'], df_sorted['entropy'], color='#E74C3C', marker='s', linewidth=1.5)
    axes[1].set_title('Shannon Entropy Profile', fontsize=12, fontweight='bold')
    axes[1].set_xlabel("Relative position from 5' end (nt)", fontsize=11)
    axes[1].set_ylabel('Shannon Entropy', fontsize=11)
    axes[1].grid(True, linestyle=':', alpha=0.6)

    plt.tight_layout()
    plt.show()
```

# Result

As a result, by analyzing mirlet7f, mirlet7d, and mirlet7f-1, able to identify points where read coverage and Shannon entropy increased sharply.

For 7f and 7f-1, found evidence of strong protein binding near the 5' end, approximately 50 nt away, while for 7d, found evidence of strong protein binding near the 40 nt mark.

Therefore, able to understand the phenomenon by which miRNA production is inhibited by protein binding.

Based on this, if miRNAs beneficial for treating a disease are not being produced due to this phenomenon, could develop a treatment strategy by designing drugs that prevent proteins from binding near the 40–50 nt region.
